```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Tạo dữ liệu mô phỏng chuẩn xác số liệu của bạn
fpr = np.linspace(0, 1, 300)
# Đường cong Champion đạt AUC xấp xỉ 0.79226
tpr_champion = 1 - (1 - fpr)**3.82
# Đường cong Baseline đạt AUC xấp xỉ 0.74546
tpr_baseline = 1 - (1 - fpr)**2.93

# 2. Khởi tạo biểu đồ phong cách Clean FinTech Dashboard
plt.figure(figsize=(7, 6), dpi=300)
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

# Đường ngẫu nhiên (Random Guess)
plt.plot([0, 1], [0, 1], color='#94A3B8', linestyle='--', linewidth=1.5, label='Random Baseline (AUC = 0.500)')

# Đường Raw Baseline
plt.plot(fpr, tpr_baseline, color='#64748B', linestyle=':', linewidth=2, label='Raw Baseline 120 feats (AUC = 0.7455)')

# Đường Champion Model
plt.plot(fpr, tpr_champion, color='#1D4ED8', linewidth=2.5, label='Champion Model 891 feats (AUC = 0.7923)')

# Đổ bóng diện tích dưới đường cong Champion
plt.fill_between(fpr, tpr_champion, color='#3B82F6', alpha=0.15, label='Shaded Area = 0.79226')

# Đánh dấu điểm cân bằng vận hành tối ưu
optimal_idx = np.argmax(tpr_champion - fpr)
opt_fpr, opt_tpr = fpr[optimal_idx], tpr_champion[optimal_idx]
plt.scatter([opt_fpr], [opt_tpr], color='#DC2626', s=50, zorder=5)
plt.annotate(f'Điểm cắt tối ưu KS\n(TPR={opt_tpr:.1%}, FPR={opt_fpr:.1%})', 
             xy=(opt_fpr, opt_tpr), xytext=(opt_fpr + 0.12, opt_tpr - 0.15),
             arrowprops=dict(facecolor='#DC2626', shrink=0.08, width=1, headwidth=6),
             fontsize=9, fontweight='semibold', color='#991B1B')

# Căn chỉnh nhãn và tiêu đề
plt.xlim([-0.01, 1.01])
plt.ylim([-0.01, 1.02])
plt.xlabel('False Positive Rate (Tỷ lệ từ chối nhầm khách tốt)', fontsize=10, fontweight='bold', color='#1E293B')
plt.ylabel('True Positive Rate (Tỷ lệ bắt trúng nợ xấu)', fontsize=10, fontweight='bold', color='#1E293B')
plt.title('TRỰC QUAN HÓA DIỆN TÍCH DƯỚI ĐƯỜNG CONG ROC (ROC-AUC)', fontsize=12, fontweight='bold', color='#0F172A', pad=12)
plt.legend(loc="lower right", frameon=True, facecolor='white', framealpha=0.95, fontsize=9)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.savefig("roc_auc_software_engineer_guide.png", dpi=300)
plt.show()
print("--> Đã xuất file thành công: roc_auc_software_engineer_guide.png")
```
